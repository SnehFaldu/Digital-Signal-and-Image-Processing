```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def log_transform(image):
    image = image.astype(np.float32)
    c = 255 / np.log(1 + np.max(image))
    result = c * np.log(1 + image)
    return np.uint8(np.clip(result, 0, 255))
def gamma_transform(image, gamma):
    normalized = image / 255.0
    result = np.power(normalized, gamma)
    return np.uint8(result * 255)
def contrast_stretch(image):
    min_value = np.min(image)
    max_value = np.max(image)
    if max_value == min_value:
        return image.copy()
    result = (image - min_value) * 255 / (max_value - min_value)
    return np.uint8(result)
def histogram_equalization(image):
    return cv2.equalizeHist(image)
def load_image(path):
    image = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
    if image is None:
        print("Unable to load:", path)
        return None
    return image
def display_images(original, enhanced, title):
    plt.figure(figsize=(10, 4))
    plt.subplot(1, 2, 1)
    plt.imshow(original, cmap="gray")
    plt.title("Original Image")
    plt.axis("off")
    plt.subplot(1, 2, 2)
    plt.imshow(enhanced, cmap="gray")
    plt.title(title)
    plt.axis("off")
    plt.tight_layout()
    plt.show()
```

```python
image1 = load_image("img_1.jpg")
if image1 is not None:
    enhanced1 = histogram_equalization(image1)
    display_images(
        image1,
        enhanced1,
        "Histogram Equalization"
    )
    cv2.imwrite("enhanced_image1.jpg", enhanced1)
if image1 is not None:
    enhanced2 = log_transform(image1)
    display_images(
        image1,
        enhanced2,
        "Log Transformation"
    )
    cv2.imwrite("enhanced_image2.jpg", enhanced2)
if image1 is not None:
    enhanced3 = gamma_transform(image1, 0.5)
    display_images(
        image1,
        enhanced3,
        "Gamma Transformation"
    )
    cv2.imwrite("enhanced_image3.jpg", enhanced3)
if image1 is not None:
    gamma_image = gamma_transform(image1, 0.5)
    enhanced4 = contrast_stretch(gamma_image)
    display_images(
        image1,
        enhanced4,
        "Gamma + Contrast Stretching"
    )
    cv2.imwrite("enhanced_image4.jpg", enhanced4)
```
