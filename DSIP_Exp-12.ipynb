```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
image = cv2.imread(r"C:\Degree Materials\Semester-5\DSIP\Experiments\img_1.jpg", 0)
if image is None:
    raise FileNotFoundError("Image not found")

_, binary_image = cv2.threshold(
    image, 127, 1, cv2.THRESH_BINARY
)
pattern = np.array([
    [-1,  1, -1],
    [ 1,  1,  1],
    [-1,  1, -1]
], dtype=np.int8)

# Perform Hit or Miss transformation
result = cv2.morphologyEx(
    binary_image,
    cv2.MORPH_HITMISS,
    pattern
)

# Display the input and processed images
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(binary_image, cmap="gray")
plt.title("Original Binary Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(result, cmap="gray")
plt.title("Hit or Miss Result")
plt.axis("off")
plt.tight_layout()
plt.show()
```
