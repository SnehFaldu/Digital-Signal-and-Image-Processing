```python
import cv2
from matplotlib import image
import numpy as np
import matplotlib.pyplot as plt
image1 = cv2.imread("img_1.jpg")
image2 = cv2.imread("img_10.jpeg")
image3 = cv2.imread("view2.jpg")
image4 = cv2.imread("view3.jpg")
image5 = cv2.imread("view4.jpg")
images = [image1, image2, image3, image4, image5]
kernel_size = (15, 15)
sigma = 1.0
gaussian_kernel = cv2.getGaussianKernel(kernel_size[0], sigma)
gaussian_kernel = np.outer(gaussian_kernel, gaussian_kernel)
def Smoothening(image): 
    Si = cv2.filter2D(image, -1, gaussian_kernel)
    return Si
sharpening_kernel = np.array([[-1, -1, -1],
                              [-1,  9, -1],
                              [-1, -1, -1]])
def Sharpening(image):    
    SHi = cv2.filter2D(image, -1, sharpening_kernel)    
    return SHi
for i in range(5):
    Og = images[i]
    Si = Smoothening(Og)
    SHi = Sharpening(Og)
    plt.figure(figsize=(15, 5))
    plt.subplot(1, 3, 1)
    plt.imshow(cv2.cvtColor(Og, cv2.COLOR_BGR2RGB))
    plt.title("Original Image")
    plt.axis('off')
    plt.subplot(1, 3, 2)
    plt.imshow(cv2.cvtColor(Si, cv2.COLOR_BGR2RGB))
    plt.title("Smoothened Image")
    plt.axis('off')
    plt.subplot(1, 3, 3)
    plt.imshow(cv2.cvtColor(SHi, cv2.COLOR_BGR2RGB))
    plt.title("Sharpened Image")
    plt.axis('off')
    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(SHi + Si , cv2.COLOR_BGR2RGB))
    plt.title("Sharpened Or Smoothened Image")
    plt.axis('off')
    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(SHi * Si , cv2.COLOR_BGR2RGB))
    plt.title("Sharpened And Smoothened Image")
    plt.axis('off')
    plt.tight_layout()
    plt.show()  
```
