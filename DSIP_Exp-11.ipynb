```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
# 1. Load the image in grayscale
image = cv2.imread("view3.jpg", 0)
if image is None:
    raise FileNotFoundError(
        "Image not found. Check the file path."
    )
# 2. Define the structuring element
kernel = np.ones((5, 5), np.uint8)
# 3. Perform erosion
erosion = cv2.erode(
    image, kernel, iterations=1
)
# 4. Perform dilation
dilation = cv2.dilate(
    image, kernel, iterations=1
)
# 5. Perform opening
opening = cv2.morphologyEx(
    image, cv2.MORPH_OPEN, kernel
)
# 6. Perform closing
closing = cv2.morphologyEx(
    image, cv2.MORPH_CLOSE, kernel
)
# 7. Display all results
titles = [
    "Original Image",
    "Erosion",
    "Dilation",
    "Opening",
    "Closing"
]
images = [
    image,
    erosion,
    dilation,
    opening,
    closing
]
plt.figure(figsize=(12, 8))
for i in range(len(images)):
    plt.subplot(2, 3, i + 1)
    plt.imshow(images[i], cmap="gray")
    plt.title(titles[i])
    plt.axis("off")
plt.tight_layout()
plt.show()
```
